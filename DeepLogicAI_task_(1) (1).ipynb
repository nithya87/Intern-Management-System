```python
def pdf_to_text(pdf_path, output_txt_path, language='deu'):
    images = convert_from_path(pdf_path)

    full_text = ""

    for i, image in enumerate(images):
        text = pytesseract.image_to_string(image, lang=language)
        full_text += text

    with open(output_txt_path, 'w', encoding='utf-8') as f:
        f.write(full_text)
```

```python
import re
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize

# Load NLTK resources
nltk.download('punkt')

def preprocess_text(file_path):
  with open(file_path, 'r', encoding='utf-8') as file:
    text = file.read()

  # Text cleaning
  text = re.sub(r'\n', ' ', text)  # Replace newline characters with space
  text = re.sub(r'\s+', ' ', text)  # Replace multiple spaces with a single space
  text = text.strip()  # Remove leading and trailing whitespace

  # Sentence segmentation
  sentences = sent_tokenize(text)

  # Tokenization
  tokenized_sentences = [word_tokenize(sentence) for sentence in sentences]

  return tokenized_sentences

preprocessed_text = []
# Example usage
for i in range(1,5):

  file_path = f'/content/drive/MyDrive/output{i}.txt'
  preprocessed_text.append(preprocess_text(file_path))

[print(preprocessed_text[i]) for i in range(4)]
```

```python
!pip install -q accelerate==0.21.0 peft==0.4.0 bitsandbytes==0.40.2 transformers==4.31.0 trl==0.4.7

import os
import torch
from datasets import load_dataset
from transformers import (
    AutoModelForCausalLM,
    AutoTokenizer,
    BitsAndBytesConfig,
    HfArgumentParser,
    TrainingArguments,
    pipeline,
    logging,
)
from peft import LoraConfig, PeftModel
from trl import SFTTrainer
```

```python
model_name = "NousResearch/llama-2-7b-chat-hf"  # or "meta-llama/Llama-2-7b-chat-hf" with Hugging Face key
dataset_name = "/content/drive/MyDrive/train.jsonl"
new_model = "llama-2-7b-custom"
lora_r = 64
lora_alpha = 16
lora_dropout = 0.1
use_4bit = True
bnb_4bit_compute_dtype = "float16"
bnb_4bit_quant_type = "nf4"
use_nested_quant = False
output_dir = "./results"
num_train_epochs = 1
fp16 = False
bf16 = False
per_device_train_batch_size = 4
per_device_eval_batch_size = 4
gradient_accumulation_steps = 1
gradient_checkpointing = True
max_grad_norm = 0.3
learning_rate = 2e-4
weight_decay = 0.001
optim = "paged_adamw_32bit"
lr_scheduler_type = "constant"
max_steps = -1
warmup_ratio = 0.03
group_by_length = True
save_steps = 25
logging_steps = 5
max_seq_length = None
packing = False
device_map = {"":0}
```

```python
train_dataset = load_dataset('json', data_files={'train': dataset_name})
valid_dataset = load_dataset('json', data_files={'validation': dataset_name})

# Preprocess datasets
train_dataset_mapped = train_dataset.map(lambda examples: {'text': [f'[INST] <<SYS>>\n{system_message.strip()}\n<</SYS>>\n\n' + prompt + ' [/INST] ' + response for prompt, response in zip(examples['prompt'], examples['response'])]}, batched=True)
valid_dataset_mapped = valid_dataset.map(lambda examples: {'text': [f'[INST] <<SYS>>\n{system_message.strip()}\n<</SYS>>\n\n' + prompt + ' [/INST] ' + response for prompt, response in zip(examples['prompt'], examples['response'])]}, batched=True)

compute_dtype = getattr(torch, bnb_4bit_compute_dtype)
bnb_config = BitsAndBytesConfig(
    load_in_4bit=use_4bit,
    bnb_4bit_quant_type=bnb_4bit_quant_type,
    bnb_4bit_compute_dtype=compute_dtype,
    bnb_4bit_use_double_quant=use_nested_quant,
)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    quantization_config=bnb_config,
    device_map=device_map
)
model.config.use_cache = False
model.config.pretraining_tp = 1
tokenizer = AutoTokenizer.from_pretrained(model_name, trust_remote_code=True)
tokenizer.pad_token = tokenizer.eos_token
tokenizer.padding_side = "right"

peft_config = LoraConfig(
    lora_alpha=lora_alpha,
    lora_dropout=lora_dropout,
    r=lora_r,
    bias="none",
    task_type="CAUSAL_LM",
)

# Set training parameters
training_arguments = TrainingArguments(
    output_dir=output_dir,
    num_train_epochs=num_train_epochs,
    per_device_train_batch_size=per_device_train_batch_size,
    per_device_eval_batch_size=per_device_eval_batch_size,
    gradient_accumulation_steps=gradient_accumulation_steps,
    optim=optim,
    save_steps=save_steps,
    logging_steps=logging_steps,
    learning_rate=learning_rate,
    weight_decay=weight_decay,
    fp16=fp16,
    bf16=bf16,
    max_grad_norm=max_grad_norm,
    max_steps=max_steps,
    warmup_ratio=warmup_ratio,
    group_by_length=group_by_length,
    lr_scheduler_type=lr_scheduler_type,
    report_to="all",
    evaluation_strategy="steps",
    eval_steps=5  # Evaluate every 5 steps
)

# Set supervised fine-tuning parameters
trainer = SFTTrainer(
    model=model,
    train_dataset=train_dataset_mapped['train'],
    eval_dataset=valid_dataset_mapped['validation'],
    peft_config=peft_config,
    dataset_text_field="text",
    max_seq_length=max_seq_length,
    tokenizer=tokenizer,
    args=training_arguments,
    packing=packing,
)

# Train the model
trainer.train()
trainer.model.save_pretrained(new_model)

# Test the model
logging.set_verbosity(logging.CRITICAL)
prompt = f"[INST] <<SYS>>\n{system_message}\n<</SYS>>\n\n ich bin harsh . [/INST]"
pipe = pipeline(task="text-generation", model=model, tokenizer=tokenizer, max_length=200)
result = pipe(prompt)
print(result[0]['generated_text'])
```

```python
import re

def extract_field(pattern, text, default="Not found"):
    match = re.search(pattern, text, re.IGNORECASE)
    return match.group(1).strip() if match else default

def analyze_german_document(document_text):
    classification = (
        "Invoice" if re.search(r"\brechnung\b", document_text, re.IGNORECASE)
        else "Unknown document type"
    )

    invoice_number = extract_field(
        r"Rechnungsnummer\s*:\s*([A-Za-z0-9-]+)", document_text
    )
    customer = extract_field(
        r"Kunde\s*:\s*([^\n\r]+)", document_text
    )
    date = extract_field(
        r"Datum\s*:\s*(\d{2}\.\d{2}\.\d{4})", document_text
    )
    product = extract_field(
        r"Produkt\s*:\s*([^\n\r]+)", document_text
    )
    amount_raw = extract_field(
        r"Gesamtbetrag\s*:\s*([\d.,]+)", document_text
    )

    if amount_raw != "Not found":
        amount = amount_raw.replace(".", "").replace(",", ".")
        amount = f"{amount} euros"
    else:
        amount = "Not found"

    details = []
    if invoice_number != "Not found":
        details.append(f"Invoice number: {invoice_number}")
    if customer != "Not found":
        details.append(f"Customer: {customer}")
    if date != "Not found":
        details.append(f"Date: {date}")
    if product != "Not found":
        details.append(f"Product: {product}")
    details.append(f"Total amount: {amount}")

    return (
        f"Classification: {classification}\n"
        f"Translation: {'; '.join(details)}\n"
        "Explanation: This invoice records transaction details "
        "and its total amount."
    )

for i in range(1, 5):
    with open(
        f"/content/drive/MyDrive/output{i}.txt",
        "r",
        encoding="utf-8"
    ) as f:
        document_text = f.read()

    print(f"\n{'=' * 15} DOCUMENT {i} {'=' * 15}")
    print(analyze_german_document(document_text))
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

model_name = "Qwen/Qwen2.5-1.5B-Instruct"

qwen_tokenizer = AutoTokenizer.from_pretrained(model_name)

qwen_model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float16
).to("cuda" if torch.cuda.is_available() else "cpu")

qwen_model.eval()

print("Qwen AI model loaded successfully!")
print("Device:", next(qwen_model.parameters()).device)
```

```python

def analyze_document_with_ai(document_text):
    messages = [
        {
            "role": "system",
            "content": (
                "You are an accurate German document understanding assistant. "
                "Analyze the supplied document using only information present in its text. "
                "If the document contains the German word 'Rechnung', classify it as 'Invoice'. "
                "Do not use the broader label 'Financial Document' for an invoice. "
                "Translate all available important fields into English, including "
                "invoice number, customer, date, product, and total amount when present. "
                "Do not invent missing information. "
                "Use exactly these three sections:\n"
                "Classification: [document type]\n"
                "Translation: [translated text and available key fields]\n"
                "Explanation: [brief explanation of the document's purpose]"
            )
        },
        {
            "role": "user",
            "content": f"Analyze this German document:\n{document_text}"
        }
    ]

    prompt = qwen_tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = qwen_tokenizer(
        prompt,
        return_tensors="pt"
    ).to(next(qwen_model.parameters()).device)

    with torch.no_grad():
        outputs = qwen_model.generate(
            **inputs,
            max_new_tokens=180,
            do_sample=False
        )

    new_tokens = outputs[0][inputs["input_ids"].shape[1]:]

    return qwen_tokenizer.decode(
        new_tokens,
        skip_special_tokens=True
    )
```

```python
ai_results = []

for i in range(1, 5):
    file_path = f"/content/drive/MyDrive/output{i}.txt"

    with open(file_path, "r", encoding="utf-8") as f:
        document_text = f.read()

    result = analyze_document_with_ai(document_text)
    ai_results.append({
        "document": f"sample{i}.pdf",
        "ocr_text": document_text,
        "analysis": result
    })

    print(f"\n{'=' * 15} DOCUMENT {i} {'=' * 15}")
    print(result)
```

```python
import pandas as pd

results_df = pd.DataFrame(ai_results)

save_path = "/content/drive/MyDrive/ai_document_analysis_results.csv"
results_df.to_csv(save_path, index=False, encoding="utf-8-sig")

print("AI results saved successfully!")
print("Location:", save_path)
display(results_df[["document", "analysis"]])
```
